**Practical-1, Revisiting pandas in numpy.**

Step-1: Importing the necessary libraries. Pandas (pd) is used to create and manage Dataframes. Numpy(np) is used for numerical and matrix operations.

```python
import pandas as pd
import numpy as np
```

Step-2: Create a dataset. The raw numerical values are converted into a Pandas Dataframe with columns A,B and C.

```python
data= [
    [3,7,2],
    [5,4,8],
    [9,6,1],
]

df = pd.DataFrame(data, columns=["A","B","C"])

print("Dataframe:")
print(df)
```

Step-3: Convert DataFrame to NumPy Array. df.to_numpy() converts the Pandas DataFrame into a NumPy array.

```python
arr = df.to_numpy()

print("\nNumPy Array:")
print(arr)
```

Step-4: Finding Shape, Size and Dimension. arr.shape shows rows and columns, arr.size shows the total number of elements, and arr.ndim shows the number of dimensions.

```python
print("\nShape:", arr.shape)
print("Size:", arr.size)
print("Dimension:", arr.ndim)
```

Step-5: Performing operations using NumPy.
1. Addition: arr + arr performs element-wise addition.

```python
addition= arr+arr
print("\nAddition:")
print(addition)
```

2. Subtraction: arr - arr performs element-wise subtraction.

```python
subtraction = arr - arr
print("\nSubtraction:")
print(subtraction)
```

3. Multiplication: Each value is multiplied by the corresponding value in the same array.

```python
multiplication = arr * arr
print("\nMultiplication:")
print(multiplication)
```

4. Transpose: Transpose flips the rows and columns of the matrix. arr.T calculates the transpose.

```python
transpose = arr.T
print("\nTranspose:")
print(transpose)
```

5. Inverse: np.linalg.inv(arr) finds the inverse of the matrix.

```python
inverse = np.linalg.inv(arr)
print("Inverse:")
print(inverse)
```

6. Product: np.dot(arr, arr) performs matrix multiplication of the array with itself.

```python
product = np.dot(arr, arr)
print("\nMatrix Product:")
print(product)
```
