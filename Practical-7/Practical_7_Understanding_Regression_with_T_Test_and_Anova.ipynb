**Practical-7 Understanding Regression with T-Test and Anova**

**Aim**:

To study the relationship between teachers' gender, age, beauty scores, and teaching evaluation scores using regression analysis, T-test, ANOVA, and correlation.

**Objectives**:

1. To understand regression analysis and its application in statistics.
2. To examine whether gender affects teaching evaluation scores using a T-test.
3. To study whether instructors' beauty scores differ according to age groups using ANOVA.
4. To examine the relationship between teaching evaluation scores and beauty scores.
5. To understand the interpretation of regression coefficients and p-values.
6. To draw conclusions from statistical results using the TeachingRatings dataset.

```python
import pandas as pd
import statsmodels.formula.api as smf
from statsmodels.stats.anova import anova_lm
```

```python
df = pd.read_csv("/content/TeachingRatings.csv")
```

```python
print(df.head())
```

**Q1. Regression with T-test**

```python
import statsmodels.formula.api as smf

# Regression model
model1 = smf.ols("eval ~ C(gender)", data=df).fit()

# Display the regression results
print(model1.summary())
```

Here:
- eval = dependent variable → teaching evaluation/rating
- gender = independent variable → male or female
- C(gender) tells Python that gender is a categorical variable.
- OLS = Ordinary Least Squares, a method used to fit a regression line/model.

Conclusion- Since the p-value is 0.001 < 0.05, we reject the null hypothesis and conclude that there is a statistically significant difference in evaluation scores based on gender.

**Q2. Regression with ANOVA**

```python
df["age_group"] = pd.cut(
    df["age"],
    bins=3,
    labels=["Young", "Middle", "Old"]
)

# Regression model
model2 = smf.ols("beauty ~ C(age_group)", data=df).fit()

# Perform ANOVA
anova_result = anova_lm(model2, typ=2)

# Display the ANOVA table
print(anova_result)
```

Here:
- beauty = dependent variable
- age_group = categorical independent variable

ANOVA = Analysis of Variance.
It checks whether the means of three or more groups are significantly different.

Conclusion-  Since the p-value is less than 0.05, we reject the null hypothesis. There is a statistically significant difference in beauty ratings among the different age groups.

Important: ANOVA tells us that a difference exists, but this output alone does not tell us exactly which two age groups differ.

```python
p_value = anova_result["PR(>F)"].iloc[0]
print("P-value:", p_value)
```

**Q3. Correlation**

```python
import statsmodels.formula.api as smf

# Regression model
model3 = smf.ols("eval ~ beauty", data=df).fit()

# Display the regression results
print(model3.summary())
```

Here:
- eval = dependent variable
- beauty = independent variable

The regression analysis shows that there is a positive and statistically significant relationship between beauty and evaluation scores. The coefficient of beauty is 0.133, with a p-value < 0.001, indicating that higher beauty ratings are associated with higher evaluation scores.

However, the R² value is only 0.036, meaning that beauty explains approximately 3.6% of the variation in evaluation scores. Therefore, although the relationship is statistically significant, its overall explanatory power is relatively weak.
